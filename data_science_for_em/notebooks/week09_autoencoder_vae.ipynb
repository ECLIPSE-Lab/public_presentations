## Part 1 — Synthetic EELS spectra: four phases with distinct peak signatures

We simulate EELS spectra near the Fe-L₂₃ edge (700–740 eV) for four iron-containing phases:

| Phase | Main edge E₀ | L₃/L₂ amplitude ratio | Physical analogy |
|---|---|---|---|
| A | 707.0 eV | 1.50 | Fe₂O₃ (Fe³⁺) |
| B | 708.5 eV | 0.80 | FeO (Fe²⁺) |
| C | 707.8 eV | 1.10 | Fe₃O₄ (mixed) |
| D | 709.2 eV | 0.50 | Surface/amorphous |

Each spectrum is the sum of two Gaussians (L₃ and L₂ edges) plus a constant background.  
We add Poisson noise (50 electrons/channel) and a small Gaussian readout noise term.

**Why this is hard for PCA:** the L₃ peak *position* varies non-linearly between phases. A PCA decomposition at rank-4 tries to fit a flat linear subspace to curved data and leaks inter-phase variance into many components.

````python
# ── Synthetic EELS spectrum generator ─────────────────────────────────────────
N = 400          # total spectra
N_PHASES = 4     # number of distinct phases
N_PER   = N // N_PHASES   # 100 spectra per phase
E = np.linspace(700, 740, 128)   # 128 energy channels

def gauss(E, E0, amp, width):
    """Single Gaussian peak."""
    return amp * np.exp(-0.5 * ((E - E0) / width) ** 2)

# Phase parameters: (L3 edge position, L3/L2 amplitude ratio, background)
PHASE_CONFIGS = [
    dict(E0=707.0, amp_ratio=1.5, bg=0.05),   # Phase A
    dict(E0=708.5, amp_ratio=0.8, bg=0.05),   # Phase B
    dict(E0=707.8, amp_ratio=1.1, bg=0.08),   # Phase C
    dict(E0=709.2, amp_ratio=0.5, bg=0.10),   # Phase D
]

clean_list = []
for cfg in PHASE_CONFIGS:
    block = []
    for _ in range(N_PER):
        E0 = cfg['E0'] + rng.normal(0, 0.15)          # small per-spectrum jitter
        amp_main = 1.0 + rng.normal(0, 0.03)
        amp_sat  = amp_main / cfg['amp_ratio'] + rng.normal(0, 0.02)
        s = (gauss(E, E0, amp_main, 1.8) +
             gauss(E, E0 + 13, amp_sat, 1.5) +
             cfg['bg'])
        block.append(np.clip(s, 0.01, None))
    clean_list.append(np.array(block, dtype=np.float32))

clean_spectra = np.vstack(clean_list)          # (400, 128)
labels_true   = np.repeat(np.arange(N_PHASES), N_PER)   # (400,)

# Add Poisson noise (50 electrons/channel) + small Gaussian readout
DOSE = 50.0
noisy_arr = rng.poisson(np.clip(clean_spectra * DOSE, 0, None)).astype(float) / DOSE
noisy_arr = np.clip(noisy_arr + rng.normal(0, 0.01, noisy_arr.shape), 0, None).astype(np.float32)

# Normalise each spectrum to its own peak for the AE (helps convergence)
# (try this yourself): what happens if you skip normalisation?
peak = noisy_arr.max(axis=1, keepdims=True) + 1e-8
X_noisy_norm = (noisy_arr / peak)
X_clean_norm = (clean_spectra / (clean_spectra.max(axis=1, keepdims=True) + 1e-8))

print(f"Dataset: {N} spectra, {N_PHASES} phases, {E.shape[0]} channels")
print(f"clean_spectra shape: {clean_spectra.shape},  noisy_arr shape: {noisy_arr.shape}")
print(f"Noisy input MSE (normalised): "
      f"{float(np.mean((X_noisy_norm - X_clean_norm)**2)):.6f}")

# ── Visualise one spectrum per phase ──────────────────────────────────────────
fig, axes = plt.subplots(1, 4, figsize=(14, 3), sharey=True)
phase_labels_str = ['A: Fe₂O₃', 'B: FeO', 'C: Fe₃O₄', 'D: Surface']
colors = ['#e74c3c', '#3498db', '#f39c12', '#2ecc71']
for k, (ax, lbl, col) in enumerate(zip(axes, phase_labels_str, colors)):
    idx = k * N_PER
    ax.plot(E, clean_spectra[idx], color=col, lw=2, label='clean')
    ax.plot(E, noisy_arr[idx],     color=col, lw=1, alpha=0.5, linestyle='--', label='noisy')
    ax.set_title(f'Phase {lbl}', fontsize=10)
    ax.set_xlabel('Energy (eV)')
    if k == 0: ax.set_ylabel('Intensity')
    ax.legend(fontsize=7)
plt.suptitle('One spectrum per phase: clean (solid) vs noisy (dashed)', fontsize=11)
plt.tight_layout()
plt.show()
````

## Part 2 — PCA baseline at k = 4 dimensions

PCA finds the best *linear* subspace of dimension $k=4$ that minimises reconstruction error.  
Because the spectral manifold is curved (peak *position* varies non-linearly between phases), PCA
needs more than 4 components to represent all four phases cleanly — some inter-phase variance leaks into higher components that we truncate.

````python
# ── PCA baseline ───────────────────────────────────────────────────────────────
K_LATENT = 4   # bottleneck dimension — same as the AE below

pca = PCA(n_components=K_LATENT, random_state=SEED)
pca_scores = pca.fit_transform(X_noisy_norm)         # (400, 4)
pca_recon  = np.clip(pca.inverse_transform(pca_scores), 0, None)   # (400, 128)

pca_recon_mse  = float(np.mean((pca_recon - X_clean_norm) ** 2))
noisy_mse_norm = float(np.mean((X_noisy_norm - X_clean_norm) ** 2))

# Cluster PCA scores with k-means
# (try this yourself): change n_init to 1 — does the silhouette change?
km_pca = KMeans(n_clusters=N_PHASES, random_state=SEED, n_init=10)
km_pca.fit(pca_scores)
sil_pca = float(silhouette_score(pca_scores, km_pca.labels_))

print(f"Noisy input MSE (normalised):  {noisy_mse_norm:.6f}")
print(f"PCA (k={K_LATENT}) recon MSE:        {pca_recon_mse:.6f}")
print(f"PCA latent silhouette score:   {sil_pca:.4f}")
print(f"Variance explained by {K_LATENT} PCs: {pca.explained_variance_ratio_.sum()*100:.1f}%")

# ── Plot PCA denoising for Phase A ────────────────────────────────────────────
idx = 0
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(E, clean_spectra[idx] / clean_spectra[idx].max(), color='#2ecc71', lw=2, label='clean (norm.)')
ax.plot(E, X_noisy_norm[idx], color='#e74c3c', lw=1, alpha=0.6, label='noisy')
ax.plot(E, pca_recon[idx], color='#f39c12', lw=2, linestyle='--', label=f'PCA k={K_LATENT}')
ax.set_xlabel('Energy (eV)'); ax.set_ylabel('Intensity (norm.)')
ax.set_title('PCA denoising: Phase A (Fe₂O₃)'); ax.legend()
plt.tight_layout(); plt.show()
````

## Part 3 — Denoising autoencoder: architecture and training

We build a three-layer MLP encoder/decoder. The bottleneck has $k=4$ dimensions — same as the PCA above.  
The model receives **noisy** spectra as input and is trained to reconstruct the **clean** spectra.  
This is the *denoising autoencoder* (DAE) introduced by Vincent et al. (2008).

**Key difference from PCA:** the encoder uses GELU nonlinearities, so the encoding function is curved.  
The AE learns to encode peak *position* and *ratio* as independent latent dimensions — a non-linear feat  
that PCA cannot achieve with the same number of components.

````python
# ── Denoising autoencoder ──────────────────────────────────────────────────────
class DenoisingAE(nn.Module):
    """Small MLP autoencoder for 1-D spectral data."""
    def __init__(self, input_dim: int = 128, latent_dim: int = 4):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 64), nn.GELU(),
            nn.Linear(64, 32),        nn.GELU(),
            nn.Linear(32, latent_dim)               # no activation on bottleneck
        )
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 32), nn.GELU(),
            nn.Linear(32, 64),         nn.GELU(),
            nn.Linear(64, input_dim),  nn.Sigmoid()  # normalised output in [0, 1]
        )

    def forward(self, x):
        z     = self.encoder(x)
        x_hat = self.decoder(z)
        return x_hat, z


def train_ae(model, X_noisy_t, X_clean_t, n_epochs=300, lr=5e-3):
    """Train a denoising AE and return final loss."""
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=n_epochs)
    model.train()
    for epoch in range(n_epochs):
        optimizer.zero_grad()
        x_hat, _ = model(X_noisy_t)
        loss = F.mse_loss(x_hat, X_clean_t)
        loss.backward()
        optimizer.step()
        scheduler.step()
    return loss.item()


# ── Train ──────────────────────────────────────────────────────────────────────
Xn_t = torch.from_numpy(X_noisy_norm)   # (400, 128)
Xc_t = torch.from_numpy(X_clean_norm)   # (400, 128)

t0 = time.time()
ae = DenoisingAE(input_dim=128, latent_dim=K_LATENT)
final_loss = train_ae(ae, Xn_t, Xc_t, n_epochs=300)
print(f"Training done in {time.time()-t0:.1f}s  |  final MSE: {final_loss:.6f}")

# ── Evaluate ───────────────────────────────────────────────────────────────────
ae.eval()
with torch.no_grad():
    x_hat_all, z_all = ae(Xn_t)

ae_recon_mse = float(F.mse_loss(x_hat_all, Xc_t).item())
z_np = z_all.numpy()                             # (400, 4) latent codes

km_ae  = KMeans(n_clusters=N_PHASES, random_state=SEED, n_init=10)
km_ae.fit(z_np)
sil_ae = float(silhouette_score(z_np, km_ae.labels_))

print(f"\n--- Results at k={K_LATENT} ---")
print(f"Noisy input MSE (normalised):  {noisy_mse_norm:.6f}")
print(f"PCA (k={K_LATENT}) recon MSE:        {pca_recon_mse:.6f}")
print(f"AE  (k={K_LATENT}) recon MSE:        {ae_recon_mse:.6f}")
print(f"PCA latent silhouette:         {sil_pca:.4f}")
print(f"AE  latent silhouette:         {sil_ae:.4f}")

# ── Self-check assertions ──────────────────────────────────────────────────────
# These checks verify the notebook's honest claims.
# They must all pass — if one fails, check SEED or n_epochs.
assert ae_recon_mse < noisy_mse_norm, (
    f"AE recon MSE ({ae_recon_mse:.6f}) should be < noisy input MSE ({noisy_mse_norm:.6f}).\n"
    "The denoising AE must outperform the raw noisy input.")

assert ae_recon_mse < pca_recon_mse, (
    f"AE recon MSE ({ae_recon_mse:.6f}) should be < PCA MSE ({pca_recon_mse:.6f}).\n"
    "The nonlinear AE should outperform PCA on the curved spectral manifold.")

assert sil_ae > 0.5, (
    f"AE latent silhouette ({sil_ae:.4f}) should be > 0.5 (well-separated clusters).")

assert sil_ae > sil_pca, (
    f"AE silhouette ({sil_ae:.4f}) should be > PCA silhouette ({sil_pca:.4f}).\n"
    "The AE latent space should cluster phases better than PCA.")

print("\nAll assertions passed.")
````

## Part 5 — Anomaly detection with reconstruction error

Train the AE on 'normal' spectra (phases A, B, C only). Then present 30 anomalous spectra  
from 'Phase D' (surface/amorphous — different peak ratio, not in the training distribution).  
Anomalies should have higher reconstruction error than normal spectra.

````python
# ── Anomaly detection demo ─────────────────────────────────────────────────────
# Train on phases A, B, C only (indices 0..299)
mask_normal   = labels_true < 3       # Phase A, B, C
mask_anomaly  = labels_true == 3      # Phase D = anomaly

ae_anom = DenoisingAE(input_dim=128, latent_dim=K_LATENT)
Xn_normal = Xn_t[mask_normal]
Xc_normal = Xc_t[mask_normal]
train_ae(ae_anom, Xn_normal, Xc_normal, n_epochs=300)

ae_anom.eval()
with torch.no_grad():
    x_hat_norm,  _ = ae_anom(Xn_t[mask_normal])
    x_hat_anom,  _ = ae_anom(Xn_t[mask_anomaly])

err_normal  = np.mean((x_hat_norm.numpy()  - X_clean_norm[mask_normal])  ** 2, axis=1)
err_anomaly = np.mean((x_hat_anom.numpy()  - X_clean_norm[mask_anomaly]) ** 2, axis=1)

threshold = np.percentile(err_normal, 99)
flagged   = (err_anomaly > threshold).sum()

print(f"Normal spectra  — mean error: {err_normal.mean():.6f}")
print(f"Anomaly spectra — mean error: {err_anomaly.mean():.6f}")
print(f"99th percentile threshold:    {threshold:.6f}")
print(f"Anomalies correctly flagged:  {flagged}/{mask_anomaly.sum()}")

# Self-check
assert err_anomaly.mean() > err_normal.mean(), (
    "Anomaly spectra should have higher mean reconstruction error than normal spectra.")
print("Anomaly detection assertion passed.")

# Plot
fig, ax = plt.subplots(figsize=(8, 4))
bins2 = np.linspace(0, err_anomaly.max() * 1.1, 30)
ax.hist(err_normal,  bins=bins2, color='#3498db', alpha=0.7, label='Normal (A,B,C)')
ax.hist(err_anomaly, bins=bins2, color='#e74c3c', alpha=0.7, label='Anomaly (D)')
ax.axvline(threshold, color='#f39c12', lw=2, linestyle='--',
           label=f'99th percentile threshold = {threshold:.4f}')
ax.set_xlabel('Reconstruction error (MSE)')
ax.set_ylabel('Count')
ax.set_title('AE anomaly detection: normal vs Phase D spectra')
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()
````

## Exercise A — How does bottleneck dimension affect reconstruction and clustering?

The lecture states: the optimal bottleneck dimension should match the intrinsic dimension of the spectral
manifold — which is close to the number of distinct phases (here: 4).

**Instructions:**
1. Run the cell below as-is to see the full sweep from $k=1$ to $k=8$.
2. Look at the two plots: reconstruction error should decrease as $k$ grows; silhouette score should peak near $k=4$ (the true number of phases).
3. **Try this yourself:** change `K_SWEEP = list(range(1, 9))` to `list(range(1, 13))` and see what happens when you over-parameterise.
4. **Try this yourself:** change `n_epochs=100` to `n_epochs=50` — does the optimal $k$ change?

## Part 6 — Variational autoencoder (VAE): a *probabilistic* latent space

The AE maps each spectrum to a **point** $z$. The VAE maps it to a **distribution** $q_\phi(z\mid x)=\mathcal{N}(\mu_\phi(x),\operatorname{diag}\sigma^2_\phi(x))$
and trains by maximising the **evidence lower bound (ELBO)**:

$$
\log p_\theta(x) \;\ge\; \underbrace{\mathbb{E}_{q_\phi(z|x)}[\log p_\theta(x\mid z)]}_{\text{reconstruction}} \;-\; \underbrace{\mathrm{KL}\big(q_\phi(z\mid x)\,\|\,\mathcal{N}(0,I)\big)}_{\text{prior matching}} .
$$

- With a Gaussian decoder of fixed variance $\sigma^2$: $-\log p_\theta(x\mid z) = \frac{1}{2\sigma^2}\|x-\hat x\|^2 + \text{const}$ — the AE's MSE, rescaled. `SIGMA2` sets the balance between the two terms.
- KL between diagonal Gaussians is closed form: $\tfrac12\sum_j(\mu_j^2+\sigma_j^2-\log\sigma_j^2-1)$.
- **Reparameterisation trick:** $z=\mu+\sigma\odot\epsilon,\ \epsilon\sim\mathcal{N}(0,I)$ — the randomness is moved off the parameter path so gradients flow through $\mu$ and $\sigma$.
- We keep the *denoising* setup (noisy input, clean target) and use a **2-D latent** so we can plot it directly, without t-SNE.
- **KL warm-up:** $\beta$ ramps from 0 to 1 over the first 50 epochs to avoid posterior collapse.

````python
# ── Variational autoencoder ────────────────────────────────────────────────────
SIGMA2 = 0.01     # decoder noise variance: smaller → reconstruction dominates the KL
LATENT_VAE = 2

class SpectralVAE(nn.Module):
    """MLP VAE for 1-D spectra: encoder outputs (mu, log sigma^2)."""
    def __init__(self, input_dim=128, latent_dim=2):
        super().__init__()
        self.body = nn.Sequential(nn.Linear(input_dim, 64), nn.GELU(),
                                  nn.Linear(64, 32),        nn.GELU())
        self.fc_mu     = nn.Linear(32, latent_dim)
        self.fc_logvar = nn.Linear(32, latent_dim)   # predict log sigma^2 → positivity for free
        self.decoder = nn.Sequential(nn.Linear(latent_dim, 32), nn.GELU(),
                                     nn.Linear(32, 64),         nn.GELU(),
                                     nn.Linear(64, input_dim),  nn.Sigmoid())

    def encode(self, x):
        h = self.body(x)
        return self.fc_mu(h), self.fc_logvar(h)

    def forward(self, x):
        mu, logvar = self.encode(x)
        eps = torch.randn_like(mu)                    # reparameterisation trick
        z = mu + torch.exp(0.5 * logvar) * eps
        return self.decoder(z), mu, logvar


def vae_loss(x_hat, x_target, mu, logvar, beta=1.0):
    """Negative ELBO per spectrum (Gaussian decoder, fixed variance SIGMA2)."""
    recon = ((x_hat - x_target) ** 2).sum(dim=1) / (2 * SIGMA2)
    kl    = 0.5 * (mu ** 2 + logvar.exp() - logvar - 1).sum(dim=1)
    return (recon + beta * kl).mean(), recon.mean(), kl.mean()


def train_vae(model, X_in, X_target, n_epochs=300, lr=3e-3, beta=1.0,
              warmup=50, batch_size=100):
    """Mini-batch training with linear KL warm-up. Returns per-epoch (recon, KL)."""
    opt = torch.optim.Adam(model.parameters(), lr=lr)
    history = []
    for ep in range(n_epochs):
        b = beta * min(1.0, (ep + 1) / warmup)       # KL warm-up
        perm = torch.randperm(X_in.shape[0])
        for i in range(0, X_in.shape[0], batch_size):
            idx = perm[i:i + batch_size]
            opt.zero_grad()
            x_hat, mu, logvar = model(X_in[idx])
            loss, rec, kl = vae_loss(x_hat, X_target[idx], mu, logvar, beta=b)
            loss.backward()
            opt.step()
        history.append((rec.item(), kl.item()))
    return np.array(history)


torch.manual_seed(SEED)
t0 = time.time()
vae = SpectralVAE(latent_dim=LATENT_VAE)
vae_hist = train_vae(vae, Xn_t, Xc_t)
print(f"VAE training done in {time.time()-t0:.1f}s")

vae.eval()
with torch.no_grad():
    mu_t, logvar_t = vae.encode(Xn_t)
    x_hat_vae = vae.decoder(mu_t)                     # decode the posterior mean
    _, rec_vae, kl_vae = vae_loss(x_hat_vae, Xc_t, mu_t, logvar_t)

mu_np    = mu_t.numpy()
sigma_np = np.exp(0.5 * logvar_t.numpy())
vae_recon_mse = float(F.mse_loss(x_hat_vae, Xc_t).item())

km_vae  = KMeans(n_clusters=N_PHASES, random_state=SEED, n_init=10).fit(mu_np)
sil_vae = float(silhouette_score(mu_np, km_vae.labels_))

print(f"Noisy input MSE:      {noisy_mse_norm:.6f}")
print(f"AE  (k=4) recon MSE:  {ae_recon_mse:.6f}")
print(f"VAE (k=2) recon MSE:  {vae_recon_mse:.6f}")
print(f"VAE mean KL per spectrum: {kl_vae.item():.2f} nats")
print(f"VAE mean posterior sigma per dim: {np.round(sigma_np.mean(0), 3)}")
print(f"k-means on VAE means: silhouette {sil_vae:.3f}, ARI vs truth {adjusted_rand_score(labels_true, km_vae.labels_):.3f}")
print(f"(for reference) ARI k-means on AE codes {adjusted_rand_score(labels_true, km_ae.labels_):.3f}, "
      f"on PCA scores {adjusted_rand_score(labels_true, km_pca.labels_):.3f}")
# Note: the VAE uses mini-batches (4 updates/epoch), the AE full-batch — the MSE values are
# not a clean architecture comparison. The point is that the VAE still denoises well.

assert vae_recon_mse < noisy_mse_norm, "The VAE must denoise: recon MSE < noisy-input MSE."
assert sil_vae > 0.5, "VAE latent means should cluster the four phases (silhouette > 0.5)."
print("VAE assertions passed.")
````

## Part 7 — GMM on the VAE latent space: BIC, soft assignments, and a trap

A Gaussian mixture model $p(z)=\sum_k \pi_k\,\mathcal{N}(z;\mu_k,\Sigma_k)$ is fitted by **EM**:

- **E-step:** responsibilities $\gamma_{ik}=\dfrac{\pi_k\mathcal{N}(z_i;\mu_k,\Sigma_k)}{\sum_j\pi_j\mathcal{N}(z_i;\mu_j,\Sigma_j)}$ (soft assignment).
- **M-step:** $\mu_k,\Sigma_k,\pi_k$ = responsibility-weighted mean, covariance and fraction.

Because the GMM has a likelihood, we can choose $K$ with the **BIC** $=-2\log p(\mathcal D;\hat\theta_K)+p_K\log N$ — something k-means cannot offer.

Then we probe the soft assignments with **interface spectra**: synthetic linear mixtures $(1-f)\,x_A+f\,x_B$ of phase A (Fe₂O₃) and phase B (FeO), as they would appear at an A/B grain boundary.

````python
# ── GMM model selection by BIC ────────────────────────────────────────────────
K_RANGE = range(1, 9)
gmms = [GaussianMixture(n_components=K, covariance_type='full',
                        random_state=SEED, n_init=3).fit(mu_np) for K in K_RANGE]
bics = np.array([g.bic(mu_np) for g in gmms])
K_BIC = int(K_RANGE[np.argmin(bics)])
gmm = gmms[K_BIC - 1]
gmm_labels = gmm.predict(mu_np)
print("BIC per K:", np.round(bics, 1))
print(f"BIC-optimal K = {K_BIC}  |  GMM ARI vs truth = {adjusted_rand_score(labels_true, gmm_labels):.3f}")

# Map each GMM component to the majority true phase (for naming only — labels never used in fitting)
comp2phase = {c: int(np.bincount(labels_true[gmm_labels == c]).argmax()) for c in np.unique(gmm_labels)}

# ── Interface spectra: A/B linear mixtures ───────────────────────────────────
fractions = np.linspace(0, 1, 21)
x_A = clean_spectra[labels_true == 0].mean(0)
x_B = clean_spectra[labels_true == 1].mean(0)
mix_clean = np.array([(1 - f) * x_A + f * x_B for f in fractions])
mix_noisy = rng.poisson(mix_clean * DOSE).astype(float) / DOSE
mix_noisy = np.clip(mix_noisy + rng.normal(0, 0.01, mix_noisy.shape), 0, None)
mix_noisy = (mix_noisy / (mix_noisy.max(1, keepdims=True) + 1e-8)).astype(np.float32)

with torch.no_grad():
    mu_mix, _ = vae.encode(torch.from_numpy(mix_noisy))
mu_mix = mu_mix.numpy()
resp_mix = gmm.predict_proba(mu_mix)                 # (21, K) responsibilities
assigned = [phase_names_str[comp2phase[c]] for c in resp_mix.argmax(1)]

print("\n f(B)   assigned phase     max responsibility")
for f, a, r in zip(fractions, assigned, resp_mix.max(1)):
    print(f" {f:4.2f}   {a:<16s}   {r:.2f}")

mid = np.argmin(np.abs(fractions - 0.5))
print(f"\n50/50 A/B interface spectrum → '{assigned[mid]}' with responsibility {resp_mix[mid].max():.2f}")

assert K_BIC == N_PHASES, f"BIC should recover the 4 simulated phases (got K={K_BIC})."
print("GMM assertion passed.")
````

````python
# ── Plot: BIC curve, GMM ellipses + interface path, responsibilities ─────────
from matplotlib.patches import Ellipse
fig, axes = plt.subplots(1, 3, figsize=(17, 4.8))

axes[0].plot(list(K_RANGE), bics, 'o-', color='#8e44ad')
axes[0].axvline(K_BIC, color='grey', ls=':')
axes[0].set_xlabel('number of components K'); axes[0].set_ylabel('BIC (lower = better)')
axes[0].set_title(f'GMM model selection: K* = {K_BIC}')

ax = axes[1]
for k in range(N_PHASES):
    m = labels_true == k
    ax.scatter(mu_np[m, 0], mu_np[m, 1], c=phase_cols[k], s=10, alpha=0.5)
for c in range(K_BIC):
    vals, vecs = np.linalg.eigh(gmm.covariances_[c])
    ang = np.degrees(np.arctan2(vecs[1, 1], vecs[0, 1]))
    ax.add_patch(Ellipse(gmm.means_[c], 2*2*np.sqrt(vals[1]), 2*2*np.sqrt(vals[0]), angle=ang,
                         fill=False, color=phase_cols[comp2phase[c]], lw=2))
ax.plot(mu_mix[:, 0], mu_mix[:, 1], 'k.-', lw=1, ms=6, label='A→B interface mixtures')
ax.set_aspect('equal'); ax.set_xlabel('μ₁'); ax.set_ylabel('μ₂'); ax.legend(fontsize=8)
ax.set_title('GMM (2σ ellipses) on VAE latent')

ax = axes[2]
for c in range(K_BIC):
    ax.plot(fractions, resp_mix[:, c], 'o-', ms=4, color=phase_cols[comp2phase[c]],
            label=phase_names_str[comp2phase[c]])
ax.set_xlabel('fraction of phase B in mixture'); ax.set_ylabel('GMM responsibility γ')
ax.set_title('Soft assignment along the A→B interface'); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
````

**The trap.** A physical 50/50 mixture of Fe₂O₃ and FeO is *not* Fe₃O₄ — yet in latent space the mixture path passes straight through the phase-C cluster, and the GMM assigns it there **with high responsibility**. Phase C's spectral fingerprint (intermediate edge position and L₃/L₂ ratio) is almost a linear blend of A and B, so the encoder cannot tell them apart.

- A responsibility is a probability *under the fitted mixture model*, not a probability that the chemistry is right.
- Clusters found without labels are hypotheses. Check them against physics: spatial context (is "phase C" only found as a thin band between A and B?), reference spectra, or a linear-unmixing fit (Week 3).

## Part 8 — t-SNE perplexity sweep: what can you read from a 2-D map?

t-SNE matches *neighbourhood probabilities* between the high-dimensional codes and a 2-D map. The **perplexity** sets the effective number of neighbours. We embed the **same** 4-D AE latent codes from Part 3 with four perplexities and measure:

1. **Rank correlation (Spearman ρ)** between the 6 inter-phase centroid distances in the 4-D latent and in the 2-D map — are "far" clusters still far?
2. **Radius ratio** (largest / smallest mean cluster radius) — are cluster *sizes* preserved?
3. **Silhouette in 2-D** using the true labels — how separated do the clusters *look*?

````python
# ── t-SNE perplexity sweep on the AE latent codes ────────────────────────────
PERPLEXITIES = [2, 5, 30, 100]      # (try this yourself): add 300 or change random_state

def centroid_stats(Y, labels):
    cents = np.array([Y[labels == k].mean(0) for k in range(N_PHASES)])
    radii = np.array([np.linalg.norm(Y[labels == k] - cents[k], axis=1).mean() for k in range(N_PHASES)])
    return cents, radii

cent_z, rad_z = centroid_stats(z_np, labels_true)
d_z = pdist(cent_z)
print(f"4-D latent: radius ratio (max/min) = {rad_z.max()/rad_z.min():.2f}")

embeds, rows = {}, []
for p in PERPLEXITIES:
    Y = TSNE(n_components=2, perplexity=p, init='pca', random_state=SEED).fit_transform(z_np)
    embeds[p] = Y
    cent_y, rad_y = centroid_stats(Y, labels_true)
    rho = spearmanr(d_z, pdist(cent_y)).correlation
    rows.append((p, rho, rad_y.max() / rad_y.min(), silhouette_score(Y, labels_true)))
    print(f"perplexity {p:>3}: Spearman ρ(distances) = {rho:5.2f}   radius ratio = {rows[-1][2]:.2f}   "
          f"2-D silhouette = {rows[-1][3]:.2f}")

fig, axes = plt.subplots(1, len(PERPLEXITIES), figsize=(18, 4.3))
for ax, (p, rho, rr, sil2) in zip(axes, rows):
    Y = embeds[p]
    for k in range(N_PHASES):
        m = labels_true == k
        ax.scatter(Y[m, 0], Y[m, 1], c=phase_cols[k], s=8, alpha=0.8)
    ax.set_title(f'perplexity {p}\nρ = {rho:.2f}, silhouette = {sil2:.2f}', fontsize=10)
    ax.set_xticks([]); ax.set_yticks([])
plt.suptitle('Same latent codes, four t-SNE maps', fontsize=11)
plt.tight_layout(); plt.show()

rhos = [r[1] for r in rows]
assert max(rhos) - min(rhos) > 0.2, "Distance ranks should change noticeably across perplexities."
print("t-SNE assertion passed.")
````

**Reading rules (write these down):**

- **Very low perplexity** (2–5) shatters real clusters into fragments and scrambles the inter-cluster distance ranking — "sub-phases" that do not exist.
- Cluster **sizes are not preserved**: t-SNE equalises densities, so a tight phase and a diffuse phase look similar. Never read area in a t-SNE plot as abundance or variability.
- **Distances between clusters** are only loosely related to distances in the latent space. Any claim "A is closer to B than to C" must be computed in $z$, not on the plot.
- A cluster is credible only if it **survives a sweep** over perplexity and random seed — and is confirmed by clustering in the original latent space (Parts 3 and 7), not in the 2-D map.
